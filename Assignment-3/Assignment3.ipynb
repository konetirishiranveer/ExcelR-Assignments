Descriptive Statistics

```python
numerical_cols = [
    'Volume',
    'Avg Price',
    'Total Sales Value',
    'Discount Rate (%)',
    'Discount Amount',
    'Net Sales Value'
]

print(df[numerical_cols].mean())
print(df[numerical_cols].median())
print(df[numerical_cols].mode().iloc[0])
print(df[numerical_cols].std())
```

interpretation

Mean: Indicates the average value of the dataset.

Median: Represents the central or middle value of the data.

Mode: Shows the most frequently occurring value.

Standard Deviation: Measures the variability or spread of the data around the mean.

Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
for col in numerical_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True)
    plt.title(f'Histogram of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

**BoxPlots**

```python
plt.figure(figsize=(10,7))

for i, col in enumerate(numerical_cols, 1):
    plt.subplot(2,3,i)
    sns.boxplot(y=df[col])
    plt.title(f'Boxplot of {col}')
    plt.ylabel(col)

plt.tight_layout()
plt.show()
```

Boxplot Interpretation

Volume: Several upper outliers are present, indicating a few transactions with very high sales
volume.

Avg Price: Contains many high-value outliers, showing that some products are priced much higher than others.
Total Sales Value: Presence of extreme values indicates some exceptionally large sales transactions.

Discount Rate (%): Fewer outliers are observed, and the distribution is relatively consistent.

Discount Amount: Several upper outliers suggest a few transactions received very high discounts.

Net Sales Value: Many high-value outliers indicate that a small number of transactions contribute significantly to total revenue.

Bar chart

```python
categorical_cols=df.select_dtypes(include='object').columns
print(categorical_cols)
```

```python
for col in categorical_cols:
    plt.figure(figsize=(8,5))
    df[col].value_counts().head(10).plot(kind='bar')

    plt.title(f'Bar Chart of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45)

    plt.show()
```

Day: Sales transactions vary across different days.

City: Certain cities contribute more sales than others.

Brand: Some brands are more popular and have higher transaction counts.

BU: Sales are concentrated in specific business units.

Model: A few models dominate sales frequency.

Conclusion

The descriptive analysis of the Sales & Discounts dataset shows that the numerical variables have different distributions and significant variability. Most numerical columns such as Volume, Avg Price, Total Sales Value, Discount Amount, and Net Sales Value are positively (right) skewed, indicating that most observations are concentrated at lower values with a few extremely high values. The Discount Rate (%) column is negatively (left) skewed.

Histograms and boxplots revealed the presence of outliers in several sales-related variables, suggesting the existence of unusually large transactions or extreme values. The categorical analysis showed variations in transaction frequencies across different categories such as cities, brands, business units, and models, indicating differences in customer preferences and market demand.
