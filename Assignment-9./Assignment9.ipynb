```python
from sklearn.preprocessing import StandardScaler
Scaler = StandardScaler()
num_cols = [
    'age',
    'fnlwgt',
    'education_num',
    'capital_gain',
    'capital_loss',
    'hours_per_week'
]
df[num_cols]=Scaler.fit_transform(df[num_cols])
```

```python
from sklearn.preprocessing import MinMaxScaler
Scaler = MinMaxScaler()
num_cols = [
    'age',
    'fnlwgt',
    'education_num',
    'capital_gain',
    'capital_loss',
    'hours_per_week'
]
df[num_cols]=Scaler.fit_transform(df[num_cols])
```

Standard Scaling is preferred when data is approximately normally distributed.

It transforms data to mean = 0 and standard deviation = 1.

Min-Max Scaling is preferred when values need to be in a fixed range, usually 0 to 1.

Min-Max Scaling is useful when a bounded range is important for the algorithm.

Both techniques make numerical features comparable in scale.

**Encoding Techniques:**

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder()
cat_cols = df.select_dtypes(include='object').columns

df[cat_cols].nunique()
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df['workclass'] = encoder.fit_transform(df['workclass'])
```

```python
df['workclass'].unique()
```

```python
df['race']=encoder.fit_transform(df['race'])
df['race'].unique()
df['native_country']=encoder.fit_transform(df['native_country'])
df['native_country'].unique()
df['capital_gain_log'] = np.log1p(df['capital_gain'])
```

One-Hot Encoding avoids creating an artificial order between categories.

Its disadvantage is that it can create many columns when there are many categories.

Label Encoding is simple and keeps the data in a single column.

Its disadvantage is that the assigned numbers may create a false order between categories.

The choice depends on the number and nature of the categorical values
