Estimation And Confidence Intervals

**A**

```python
import numpy as np
```

```python
data = [1.13, 1.55, 1.43, 0.92, 1.25,
1.36, 1.32, 0.85, 1.07, 1.48,
1.20, 1.33, 1.18, 1.22, 1.29]
n = len(data)
mean = np.mean(data)
print(n)
print(mean)
```

```python
std=np.std(data, ddof=1)
print(std)
```

```python
from scipy.stats import t

confidence = 0.99
alpha =1 - confidence
df = n-1
t_critical= t.ppf(1-alpha/2,df)
print(t_critical)
```

```python
ME = t_critical *std / np.sqrt(n)
```

```python
lower = mean - ME
upper = mean + ME
print(lower)
print(upper)
```
